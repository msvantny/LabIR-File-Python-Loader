## 2. Load the thermography recording

`load_lrc` reads the entire recording through the bundled DLL and prints basic metadata. It returns `frames`, a `float32` NumPy array with shape **(height, width, number of frames)**, and `times`, a `float64` array of frame times in seconds. Both arrays can be used directly for further processing.

The included example has 6 frames at 320 × 256 pixels. Replace `file_path` with a relative or absolute path to your own LRC file. Large recordings require enough RAM for the entire array. LRC defines stream-unit metadata, but the DLL C wrapper used here does not return the unit code. Values may be temperature, intensity, or another camera-dependent quantity; do not assume degrees Celsius. This Python layer applies no additional unit conversion or radiometric correction. Confirm the unit from explicit recording metadata and camera/export documentation. The setup guides link the public library, C API header, format specification, and pure-Python alternative.

The path below is relative to the project folder. You can also use an absolute Windows path, for example `r"D:\Measurements\my_recording.lrc"`.

**Path limitation:** use ASCII-only file and folder names throughout the full LRC path. Accented names such as `měření.lrc` failed in the current DLL-based loader; a raw Python string (`r"..."`) does not fix this limitation. Spaces in file names worked.

```python
from loaders import load_lrc
from visualization import show_sequence_slider

# Replace this path to load another recording.
file_path = "LRC_Files/sample_sequence_320x256_250Hz_006Fr.lrc"
frames, times = load_lrc(file_path)
```

## 3. Browse the temperature field

The **Frame index** slider selects a frame from the recording. The image shows the spatial distribution of temperature values, while the color bar on the right maps each color to a numerical value. The title gives the current frame, its position in the sequence, and its time from the beginning of the recording. The color scale dynamically follows the minimum and maximum of the current frame, making small spatial differences easier to see. Because the scale changes between frames, always check the color-bar values when comparing them. This viewer does not select points with the mouse or create temporal-signal plots.

```python
# Display the sequence with actual frame times in seconds.
viewer = show_sequence_slider(frames, time_axes=times)
```
